```python
from transformers import pipeline

# Chargement du modèle de base demandé dans le ticket
print("Chargement du modèle DistilBERT...")
classifier = pipeline("text-classification", model="distilbert-base-uncased")
print("Modèle chargé avec succès !")
```

```python
# Simulation de textes issus du dataset
emails_test = [
    "Hello team, please find attached the weekly report for our marketing campaign. Best, John.", # Mail normal (Ham)
    "URGENT: Your bank account has been suspended. Click here immediately to verify your identity and restore access." # Phishing (Spam)
]

# Analyse par l'IA
for email in emails_test:
    resultat = classifier(email)
    print(f"Texte : {email}")
    print(f"Prédiction brute : {resultat}\n")
```
